1. (6б). Створіть клас Building, додайте три атрибути, один з яких має значення за замовчуванням та два методи на свій розсуд.

```python
class Building:
    def __init__(self, name, year_built, material='Brick'):
        self.name = name
        self.year_built = year_built
        self.material = material

    def get_age(self):
        from datetime import datetime
        current_year = datetime.now().year
        return current_year - self.year_built

    def display_info(self):
        return f"Building Name: {self.name}, Year Built: {self.year_built}, Material: {self.material}"
```

3. (5б). Створіть клас, де атрибути мають різні рівні доступу. Спробуйте отримати їхні значення та опишіть результати.

```python
class AccessDemo:
    def __init__(self, public_attr, protected_attr, private_attr):
        self.public_attr = public_attr          # Public attribute
        self._protected_attr = protected_attr  # Protected attribute (by convention)
        self.__private_attr = private_attr      # Private attribute (name mangling)

    def display_attrs(self):
        print(f"Public: {self.public_attr}")
        print(f"Protected: {self._protected_attr}")
        print(f"Private (via method): {self.__private_attr}")

# Create an object
demo_obj = AccessDemo("I am public", "I am protected", "I am private")

# Accessing attributes
print("\n--- Accessing Attributes Directly ---")
print(f"Accessing public_attr directly: {demo_obj.public_attr}")

# Accessing protected attribute (by convention, still accessible directly)
print(f"Accessing _protected_attr directly: {demo_obj._protected_attr}")

# Attempting to access private attribute directly (will cause AttributeError if uncommented)
# print(f"Accessing __private_attr directly: {demo_obj.__private_attr}")
# print("Direct access to __private_attr causes an AttributeError.")

# Accessing private attribute via name mangling
print(f"Accessing __private_attr via name mangling: {demo_obj._AccessDemo__private_attr}")

print("\n--- Accessing Attributes via Class Method ---")
demo_obj.display_attrs()
```

5. (9б). Створіть клас 3D-фігура без атрибутів, з методом get_volume для отримання об'єму фігури, що повертає 0 та __add __, який приймає self та other в якості аргументів, а повертає суму об'ємів 3d-фігур self та other.

```python
class ThreeDShape:
    def get_volume(self):
        return 0

    def __add__(self, other):
        if isinstance(other, ThreeDShape):
            return self.get_volume() + other.get_volume()
        else:
            raise TypeError("Unsupported operand type for +: 'ThreeDShape' and " + type(other).__name__)
```

6. (11б). Створіть 2 дочірніх класи від 3d-фігура: Циліндр та Куб, які мають атрибути, необхідні для розрахунку об'єму. Визначте метод get_volume в кожному з них так, щоби вони розраховували об'єм у залежності від формули для кожного типу фігури. Створіть обʼєкт класу Циліндр та обʼєкт класу Куб. Виконайте операцію суми за допомогою оператора + між ними, аби знайти сумарний об'єм.

```python
import math

class Cylinder(ThreeDShape):
    def __init__(self, radius, height):
        self.radius = radius
        self.height = height

    def get_volume(self):
        return math.pi * self.radius**2 * self.height

class Cube(ThreeDShape):
    def __init__(self, side):
        self.side = side

    def get_volume(self):
        return self.side**3

cylinder_obj = Cylinder(radius=2, height=5)
cube_obj = Cube(side=3)

print(f"Volume of Cylinder: {cylinder_obj.get_volume():.2f}")
print(f"Volume of Cube: {cube_obj.get_volume():.2f}")

total_volume = cylinder_obj + cube_obj
print(f"Total volume (Cylinder + Cube): {total_volume:.2f}")
```

7. (3б). Продемонструйте різницю між isinstance та issubclass.

```python
class Parent:
    pass

class Child(Parent):
    pass

parent_obj = Parent()
child_obj = Child()

print("--- isinstance examples ---")
print(f"Is child_obj an instance of Child? {isinstance(child_obj, Child)}")
print(f"Is child_obj an instance of Parent? {isinstance(child_obj, Parent)}")
print(f"Is parent_obj an instance of Child? {isinstance(parent_obj, Child)}")
print(f"Is parent_obj an instance of Parent? {isinstance(parent_obj, Parent)}")
print(f"Is 'hello' an instance of str? {isinstance('hello', str)}")
print(f"Is 123 an instance of int? {isinstance(123, int)}")

print("\n--- issubclass examples ---")
print(f"Is Child a subclass of Parent? {issubclass(Child, Parent)}")
print(f"Is Parent a subclass of Child? {issubclass(Parent, Child)}")
print(f"Is Child a subclass of Child? {issubclass(Child, Child)}") # A class is considered a subclass of itself
print(f"Is int a subclass of object? {issubclass(int, object)}")
print(f"Is str a subclass of int? {issubclass(str, int)}")
```

8. (13б). Створіть клас EWallet з приватними атрибутами balance та id. Реалізуйте методи поповнення та зняття коштів, забезпечивши належну інкапсуляцію. Підказка: використовуйте декоратори property та setter.

```python
class EWallet:
    def __init__(self, initial_balance, wallet_id):
        self.__balance = initial_balance
        self.__id = wallet_id

    @property
    def balance(self):
        return self.__balance

    @property
    def wallet_id(self):
        return self.__id

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Deposited {amount}. New balance: {self.__balance}")
        else:
            print("Deposit amount must be positive.")

    def withdraw(self, amount):
        if amount > 0:
            if self.__balance >= amount:
                self.__balance -= amount
                print(f"Withdrew {amount}. New balance: {self.__balance}")
            else:
                print("Insufficient funds.")
        else:
            print("Withdrawal amount must be positive.")

wallet = EWallet(initial_balance=100, wallet_id="user123")

print(f"Initial Balance: {wallet.balance}")
print(f"Wallet ID: {wallet.wallet_id}")

wallet.deposit(50)
wallet.withdraw(30)
wallet.withdraw(150)
wallet.deposit(-10)

print(f"Final Balance: {wallet.balance}")
```

9. (11б). Створіть клас ShoppingCart, який містить список об'єктів типу Item. Реалізуйте методи для додавання айтемів, їх видалення та відображення списку айтемів.

```python
class Item:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def __str__(self):
        return f"{self.name} (${self.price:.2f})"

class ShoppingCart:
    def __init__(self):
        self.items = []

    def add_item(self, item, quantity=1):
        for _ in range(quantity):
            self.items.append(item)
        print(f"Added {quantity} x {item.name}(s) to the cart.")

    def remove_item(self, item_name):
        found = False
        for item in list(self.items):
            if item.name == item_name:
                self.items.remove(item)
                found = True
                print(f"Removed one {item_name} from the cart.")
                break
        if not found:
            print(f"Item '{item_name}' not found in the cart.")

    def display_items(self):
        if not self.items:
            print("Shopping cart is empty.")
            return
        print("\n--- Shopping Cart Items ---")
        for item in self.items:
            print(f"- {item}")
        total = sum(item.price for item in self.items)
        print(f"---------------------------")
        print(f"Total: ${total:.2f}")

print("\n--- Demonstrating ShoppingCart ---")
cart = ShoppingCart()

apple = Item("Apple", 0.50)
banana = Item("Banana", 0.75)
orange = Item("Orange", 1.20)

cart.add_item(apple, 3)
cart.add_item(banana)
cart.add_item(orange, 2)

cart.display_items()

cart.remove_item("Apple")
cart.remove_item("Grape")

cart.display_items()

cart.remove_item("Banana")
cart.remove_item("Orange")
cart.remove_item("Orange")

cart.display_items()
```

10. (13б). Створіть клас Vehicle з атрибутами brand (manufacturer) та production_age. Створіть ще один клас Engine з такими атрибутами, як type (ex. petrol) та size (volume, litres). Створіть клас Car, який успадковує обидва класи Vehicle та Engine. Продемонструйте використання множинної спадковості, створивши обʼєкт класу Car та отримавши mro для цього класу.

```python
class Vehicle:
    def __init__(self, brand, production_age):
        self.brand = brand
        self.production_age = production_age

class Engine:
    def __init__(self, engine_type, size):
        self.engine_type = engine_type
        self.size = size

class Car(Vehicle, Engine):
    def __init__(self, brand, production_age, engine_type, size):
        Vehicle.__init__(self, brand, production_age)
        Engine.__init__(self, engine_type, size)

    def __str__(self):
        return f"{self.brand} ({self.production_age} years old) with {self.size}L {self.engine_type} engine"


my_car = Car("Toyota", 5, "petrol", 2.5)
print(f"Car Info: {my_car}")


print("\nMethod Resolution Order (MRO):")
print(Car.mro())
```

Ітерабельний об'єкт — це колекція, яку можна обійти (наприклад, список або рядок), тоді як ітератор — це спеціальний об'єкт зі станом, який пам'ятає поточну позицію і повертає наступний елемент при виклику `next()`.

13. (11б). Створіть клас, що має визначений Iterator Protocol та при кожному виклику next повертає назву дня тижня, поки вони не вичерпаються. Зверніть увагу, що порядок має значення.

```python
class WeekdayIterator:
    def __init__(self):
        self.days = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]
        self.index = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.index < len(self.days):
            day = self.days[self.index]
            self.index += 1
            return day
        else:
            raise StopIteration

week_iter = WeekdayIterator()
for day in week_iter:
    print(day)
```
